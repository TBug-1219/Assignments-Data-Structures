```python
array = [6, 4, 12, 9, 45, 23, 2, 43, 56, 17, 65, 11, 7]

smallestPair = array[0], array[1]
largestPair = array[0], array[1]

smallestSum = array[0] + array[1]
largestSum = array[0] + array[1]

for ch in range(len(array)-1):
    currentSum = array[ch]+ array[ch + 1]

    if currentSum < smallestSum:
        smallestSum = currentSum
        smallestPair = (array[ch], array[ch + 1])

    if currentSum > largestSum:
        largestSum = currentSum
        largestPair = (array[ch], array[ch + 1])

print(f"Smallest value pair in array are {smallestPair} for a total of {smallestSum}")
print(f"Largest value pair in array are {largestPair} for a total of {largestSum}")
```

```python
list1 = ['a','b', 'g', 'h', 't']
list2 = ['c', 'd', 'f', 'b', 't']


commonLetters = set(list1).intersection(list2)

print("Common Letters:", commonLetters)

for letter in commonLetters:
    pos1 = list1.index(letter)
    pos2 = list2.index(letter)
    print(f"Letter '{letter}' found at position {pos1} in List1 and position {pos2} in List2.")
```

```python
list1 = {1: 'a', 2: 'b', 3: 'c', 4:'d', 5: 'e'}
list2 = {'a': 'U.S.', 'b': 'Brazil', 'c': 'India',
         'd': 'France', 'e': 'Ireland'}

number = int(input("Enter a number 1-5: "))

if number in list1:
    letter = list1[number]

    country = list2.get(letter, "Unknown Country")
    print(f"Number {number}  is associated with the letter '{letter}' and country '{country}'.")

else:
    print("Invalid input. Please enter a number 1-5.")
```
