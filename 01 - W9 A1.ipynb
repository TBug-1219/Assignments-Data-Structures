```python
import random
import time
```

```python
#Bubble Sort Sorted 100
arrayS100 = list(range(1,101))
arrayRU100 = []

start_Time = time.time()
print("In Order: ")
for i in range(0, len(arrayS100), 30):
    print(arrayS100[i:i+30])

def bubble_Sort(arrayS100):
    for n in range(len(arrayS100) - 1, 0, -1):
        swapped = False
        for i in range(n):
            if arrayS100[i] > arrayS100[i+1]:
                arrayS100[i], arrayS100[i+1] = arrayS100[i+1], arrayS100[i]
                swapped = True
        if not swapped:
            break
bubble_Sort(arrayS100)
print("After Sort: ")
for i in range (0, len(arrayS100), 30):
    print(arrayS100[i:i+30])
end_Time = time.time()
elapsed_Time = end_Time - start_Time
print(f"Elapsed Time = {elapsed_Time}")
```

```python
#Merge Sort Sorted 100
arrayS100 = list(range(1,101))
arrayRU100 = []

start_Time = time.time()
print("In Order: ")
for i in range(0, len(arrayS100), 30):
    print(arrayS100[i:i+30])

def merge(arrayS100, l, m, r):
    n1 = m - l + 1
    n2 = r - m
    L = [0] * (n1)
    R = [0] * (n2)

    for i in range(0, n1):
        L[i] = arrayS100[l+i]
    for j in range(0, n2):
        R[j] = arrayS100[m + 1 + j]

    i = 0
    j = 0
    k = l

    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arrayS100[k] = L[i]
            i += 1
        else:
            arrayS100[k] = R[j]
            j += 1
        k += 1

    while i < n1:
        arrayS100[k] = L[i]
        i += 1
        k +=1
    
    while j < n2:
        arrayS100[k] = R[j]
        j += 1
        k += 1

def merge_Sort(arrayS100, l, r):
    if l < r:
        m = l+(r-l)//2
        merge_Sort(arrayS100, l, m)
        merge_Sort(arrayS100, m+1, r)
        merge(arrayS100, l, m, r)
        
n = len(arrayS100)
merge_Sort(arrayS100, 0, n-1)
print("After Sort: ")
for i in range (0, len(arrayS100), 30):
    print(arrayS100[i:i+30])
    
end_Time = time.time()
elapsed_Time = end_Time - start_Time
print(f"Elapsed Time = {elapsed_Time}")
```

```python
#Quick Sort Sorted 100
arrayS100 = list(range(1,101))
arrayRU100 = []

start_Time = time.time()
print("In Order: ")
for i in range(0, len(arrayS100), 30):
    print(arrayS100[i:i+30])

def partition(arrayS100, low, high):
    pivot = arrayS100[high]
    i = low -1
    for j in range(low, high):
        if arrayS100[j] <= pivot:
            i = i + 1
            (arrayS100[i], arrayS100[j]) = (arrayS100[j], arrayS100[i])
    (arrayS100[i + 1], arrayS100[high]) = (arrayS100[high], arrayS100[i + 1])
    return i + 1

def quick_Sort(arrayS100, low, high):
    if low < high:
        pi = partition(arrayS100, low, high)
        quick_Sort(arrayS100, low, pi - 1)
        quick_Sort(arrayS100, pi + 1, high)

quick_Sort(arrayS100, 0, len(arrayS100) -1)
print("After Sort: ")
for i in range (0, len(arrayS100), 30):
    print(arrayS100[i:i+30])
    
end_Time = time.time()
elapsed_Time = end_Time - start_Time
print(f"Elapsed Time = {elapsed_Time}")
```

```python
#Merge Sort Sorted 10000
arrayS10000 = list(range(1,10001))
arrayRU10000 = []

start_Time = time.time()
print("In Order: ")
for i in range(0, len(arrayS10000), 30):
    print(arrayS10000[i:i+30])

def merge(arrayS10000, l, m, r):
    n1 = m - l + 1
    n2 = r - m
    L = [0] * (n1)
    R = [0] * (n2)

    for i in range(0, n1):
        L[i] = arrayS10000[l+i]
    for j in range(0, n2):
        R[j] = arrayS10000[m + 1 + j]

    i = 0
    j = 0
    k = l

    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arrayS10000[k] = L[i]
            i += 1
        else:
            arrayS10000[k] = R[j]
            j += 1
        k += 1

    while i < n1:
        arrayS10000[k] = L[i]
        i += 1
        k +=1
    
    while j < n2:
        arrayS10000[k] = R[j]
        j += 1
        k += 1

def merge_Sort(arrayS10000, l, r):
    if l < r:
        m = l+(r-l)//2
        merge_Sort(arrayS10000, l, m)
        merge_Sort(arrayS10000, m+1, r)
        merge(arrayS10000, l, m, r)

merge_Sort(arrayS10000, 0, len(arrayS10000)-1)
print("After Sort: ")
for i in range (0, len(arrayS10000), 30):
    print(arrayS10000[i:i+30])
    
end_Time = time.time()
elapsed_Time = end_Time - start_Time
print(f"Elapsed Time = {elapsed_Time}")
```

```python
#Quick Sort Sorted 10000
arrayS10000 = list(range(1,10001))
arrayRU10000 = []

start_Time = time.time()
print("In Order: ")
for i in range(0, len(arrayS10000), 30):
    print(arrayS10000[i:i+30])

def partition(arrayS10000, low, high):
    pivot = arrayS10000[high]
    i = low -1
    for j in range(low, high):
        if arrayS10000[j] <= pivot:
            i = i + 1
            (arrayS10000[i], arrayS10000[j]) = (arrayS10000[j], arrayS10000[i])
    (arrayS10000[i + 1], arrayS10000[high]) = (arrayS10000[high], arrayS10000[i + 1])
    return i + 1

def quick_Sort(arrayS10000, low, high):
    if low < high:
        pi = partition(arrayS10000, low, high)
        quick_Sort(arrayS10000, low, pi - 1)
        quick_Sort(arrayS10000, pi + 1, high)

quick_Sort(arrayS10000, 0, len(arrayS10000) -1)
print("After Sort: ")
for i in range (0, len(arrayS10000), 30):
    print(arrayS10000[i:i+30])
    
end_Time = time.time()
elapsed_Time = end_Time - start_Time
print(f"Elapsed Time = {elapsed_Time}")
```
